# Django + OpenAI + RAG (Devam Seviye)

Bu dokuman kod vermez.
Sadece ogrencinin adim adim ne yapacagini anlatir.

Tema: Dokuman tabanli cevap ureten bir RAG akisinin kurulmasi.

## 1. Bu Fazin Amaci

Bu calismanin amaci, ogrencinin su becerileri kazanmasidir:
- RAG mantigini uc ana adimda kavramak (index, retrieval, generation)
- LangChain ile retriever tabanli akis kurmak
- Dokumanlardan kanitli cevap uretmek
- Cevapta kaynak gostermeyi standartlastirmak

Beklenen cikti: Uygulama, sorulara dokumana dayali cevap verebilir hale gelir.

## 2. Baslamadan Once

Ogrenci asagidakileri hazirlar:
- Projedeki dokuman dosyalari (README ve ilgili notlar)
- OpenAI API anahtari
- LangChain ve embedding kutuphaneleri
- Vektor veritabani secimi (FAISS veya Chroma)

Beklenen cikti: RAG kurulumuna gecmek icin teknik zemin hazir olur.

## 3. RAG Altyapisini Kur

Ogrenci su sira ile ilerler:
1. Dokumanlari oku ve temizle.
2. Metni parcalara bol (chunk) ve boyutlari belirle.
3. Embedding olustur.
4. Vektor index yaz ve kaydet.
5. Retriever ile ilgili parcalari getir.
6. LLM'e baglam + soru vererek yanit uret.

Beklenen cikti: Soruya ilgili dokuman parcasi ile yanit ureten temel RAG akisi olusur.

## 4. Klasor ve Sorumluluklari Belirle

Ogrenci proje icinde su rollerin oldugu bir duzen kurar:
- `rag/loader`: dokuman alma ve temizleme
- `rag/indexer`: chunk + embedding + index yazma
- `rag/retriever`: ilgili parcayi bulma
- `rag/chain`: prompt + llm + kaynak donus akisi
- API katmani: soruyu alip RAG cikisini donen nokta

Beklenen cikti: Her sorumluluk ayri dosya/bolumde net ayrilmis olur.

## 5. Kalite ve Dogrulama

Ogrenci su kontrolleri uygular:
1. Cevap en az bir dokuman kaynagi iceriyor mu?
2. Ayni soruda baglamsiz modele gore daha tutarli sonuc veriyor mu?
3. Dokuman degisince index yenilenip cevap degisiyor mu?
4. Hata durumunda kullaniciya anlasilir mesaj donuluyor mu?

Beklenen cikti: RAG akisinin dogruluk ve guvenirlik seviyesi gorulebilir hale gelir.

## 6. Faz Sonu Beklenen Cikti

Bu faz sonunda sistem:
- Dokumanlardan baglam cekerek cevap uretir
- Cevapta kaynak bilgisini gosterir
- Temel hata durumlarini yonetir

Bir sonraki faz: MCP ile arac katmanini standardize etmek.
