# Django + OpenAI Function Calling (Baslangic Seviye)

Bu dokuman kod vermez.
Sadece ogrencinin adim adim ne yapacagini anlatir.

Tema: Lokal klasorden proje alma ve kod duzenleme sureci.

## 1. Projenin Amaci

Bu calismanin amaci, ogrencinin su becerileri kazanmasidir:
- Django ile API ve on yuzu birlikte kurgulamak
- On yuzde Django Template kullanarak basit bir chat/komut ekrani hazirlamak
- OpenAI function calling mantigini anlamak
- Modelin arac cagirdigi bir is akisinda dogru sirayla ilerlemek
- Lokal kaynaktan proje alma, dosya okuma, metin degistirme adimlarini yonetmek
- Uygulamayi Dockerfile ve Docker Compose ile calisir hale getirmek
- Kritik adimlari uygulama loglarindan izleyebilmek

## 2. Baslamadan Once

Ogrenci asagidakileri hazirlar:
- Guncel bir Python surumu
- OpenAI API anahtari
- Calisma klasoru
- Icinden proje alinacak lokal kaynak klasor
- Docker ve Docker Compose

Beklenen cikti: Ortam hazir ve proje kuruluma gecilebilir.

## 3. Proje Iskeletini Kur

Ogrenci su sira ile ilerler:
1. Yeni bir proje klasoru acar.
2. Sanal ortam olusturur ve aktif eder.
3. Django ve OpenAI kutuphanelerini kurar.
4. Django projesi ve tek bir uygulama olusturur.

Beklenen cikti: Django projesi acilabilir durumda olur.

## 4. Yapisal Klasorleri Belirle

Ogrenci proje icinde su rollerin oldugu bir duzen kurar:
- API katmani (istek alma ve cevap donme)
- Agent katmani (LLM ile konusma ve function call dongusu)
- Tool katmani (lokal proje alma, dosya okuma, metin degistirme)
- Template katmani (kullanici ekranlari)
- Static katmani (stil ve istemci tarafi varliklar)
- Calisma alani klasoru (alinan projelerin tutuldugu guvenli alan)

Beklenen cikti: Her sorumluluk ayri dosya/bolumde net ayrilmis olur.

## 5. Ayarlar ve Anahtarlar

Ogrenci su prensipleri uygular:
1. API anahtari dogrudan kodun icine yazilmaz.
2. Ortam degiskeni dosyasi kullanilir.
3. Django ayarlarinda model adi ve anahtar okunur.
4. Uygulama, INSTALLED_APPS listesine eklenir.
5. Template ve static dizin ayarlari acikca yapilir.

Beklenen cikti: Servis anahtarlari guvenli ve merkezden yonetilir.

## 6. Tool Tasarimini Yap

Ogrenci 3 temel araci tanimlar (kodsuz tasarim):
1. Lokal bir kaynak klasorden projeyi calisma alanina alma araci
2. Metin dosyasini okuma araci
3. Dosya iceriginde bir metni digeriyle degistirme araci

Her arac icin su net yazilir:
- Aracin amaci
- Hangi girdileri bekledigi
- Basarili sonuc formati
- Hata durumunda donulecek mesaj mantigi

## 7. Guvenlik Kurallari

Ogrenci asagidaki minimum kurallari uygular:
- Yalnizca belirlenen calisma klasoru altinda duzenleme yapilir
- Yalnizca izinli lokal kaynak klasorden proje alinmasina izin verilir
- Ust dizine cikis denemeleri (or. ../) reddedilir
- Olmayan dosya/klasor icin acik hata mesaji donulur
- Ayni hedef klasore ikinci kez proje alma denemesi kontrol edilir

Beklenen cikti: Araclar kontrolsuz dosya islemine izin vermez.

## 8. Function Calling Akisini Kur

Ogrenci bu is akis mantigini uygular:
1. Kullanici mesaji modele gider.
2. Model gerekirse tool cagrisi onerir.
3. Sistem ilgili tool'u calistirir.
4. Tool sonucu modele geri verilir.
5. Model nihai cevabi uretir.

Ek kural: Sonsuz donguyu engellemek icin maksimum tur limiti tanimlanir.

## 9. Django API Ucu

Ogrenci tek bir endpoint tasarlar:
- Yalnizca POST kabul eder
- JSON icinde kullanici mesajini alir
- Agent akisini tetikler
- Sonucu yine JSON olarak doner
- Istek, tool cagrisi ve sonuc adimlarini loglara yazar

Hata senaryolari icin net durumlar yazilir:
- Bos mesaj
- Gecersiz JSON
- Ic islem hatasi

## 10. On Yuz: Django Template

Ogrenci on yuz tarafinda su adimlari tamamlar:
1. Template dosyalari icin duzenli klasor yapisi kurar.
2. Kullanici komutunu yazabilecegi basit bir ekran tasarlar.
3. Form veya buton ile API endpointine istek gonderme akisini kurar.
4. Donen cevabi ayni ekranda gorunur hale getirir.
5. Hata mesajlarini kullaniciya okunur sekilde gosterir.

Beklenen cikti: Tarayicida calisan, Django Template tabanli temel bir arayuz bulunur.

## 11. Dockerfile ve Docker Compose

Ogrenci dagitim adimlarini su sekilde kurgular:
1. Uygulamayi konteynerde calistiracak Dockerfile hazirlar.
2. Servisi calistirmak icin docker-compose dosyasi olusturur.
3. Ortam degiskenlerinin container icinde dogru yuklendigini kontrol eder.
4. Uygulamayi compose ile ayaga kaldirip endpoint ve template ekranini test eder.
5. Uygulama loglarini ve container loglarini okuyup akisin gorundugunu dogrular.

Beklenen cikti: Proje tek komutla konteyner uzerinden calisabilir hale gelir.

## 12. Ogrenci Test Senaryolari

Ogrenci asagidaki senaryolari tek tek dener:
1. Lokal kaynak klasorden projeyi yeni calisma klasorune alma
2. Alinan projeden bir metin dosyasi okuma
3. Okunan dosyada hedef bir metni degistirme
4. Olmayan dosyayi okuma denemesi
5. Guvensiz path ile erisim denemesi
6. Django Template ekranindan API akisinin calismasi
7. Docker Compose ile uygulamanin basariyla kalkmasi
8. Yapilan bir islemde tool cagrilarinin ve sonucun loglarda gorunmesi

Her senaryoda beklenen sonuc: Basari veya hata mesaji anlasilir sekilde gorulmeli.
