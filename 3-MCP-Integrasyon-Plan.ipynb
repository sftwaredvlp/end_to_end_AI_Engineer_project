# Django + OpenAI + MCP (Integrasyon Seviye)

Bu dokuman kod vermez.
Sadece ogrencinin adim adim ne yapacagini anlatir.

Tema: Agent araclarini MCP protokolu ile standardize etme sureci.

## 1. Bu Fazin Amaci

Bu calismanin amaci, ogrencinin su becerileri kazanmasidir:
- MCP'nin neden gerekli oldugunu anlamak
- Arac cagrilarini standart protokol ile yonetmek
- Agent katmaninda arac bagimliliklarini sadelemek
- Tool guvenligi ve erisim sinirlarini netlestirmek

Beklenen cikti: Arac ekosistemi daha olceklenebilir ve izlenebilir hale gelir.

## 2. Baslamadan Once

Ogrenci asagidakileri hazirlar:
- Calisan bir RAG + LangChain akisinin temel surumu
- Mevcut tool listesinin net dokumu
- Guvenli bir calisma klasoru siniri
- Loglama stratejisi (kim, neyi, ne zaman)

Beklenen cikti: MCP gecisi icin teknik ve operasyonel zemin hazir olur.

## 3. MCP Tool Tasarimini Yap

Ogrenci su sira ile ilerler:
1. Ilk tool setini belirler (kopyalama, okuma, degistirme, listeleme).
2. Her tool icin giris dogrulama kurallarini yazar.
3. Cikis formatini standartlastirir.
4. Hata mesaji formatini tek tip hale getirir.
5. Yetki sinirlarini belirler.

Beklenen cikti: MCP uzerinden cagirilabilir, tutarli bir tool katmani olusur.

## 4. API ve Agent Entegrasyonunu Kur

Ogrenci su adimlari uygular:
- Django API istegi alir ve niyeti belirler
- Agent uygun MCP tool cagrisini secer
- MCP sonucu standart formatta doner
- Sonuc LLM yanitina baglanir ve kullaniciya sunulur

Beklenen cikti: Uygulama, MCP araclarini dogru sirayla kullanabilir hale gelir.

## 5. Guvenlik ve Izlenebilirlik

Ogrenci su prensipleri uygular:
1. Toollar yalnizca izinli klasorde calisir.
2. Path traversal benzeri riskler engellenir.
3. Her cagrida giris/cikis loglari tutulur.
4. Hatalar kullaniciya acik ama guvenli bicimde donulur.

Beklenen cikti: Sistem daha guvenli ve denetlenebilir hale gelir.

## 6. Basari Kriterleri

- En az 3 temel tool MCP uzerinden sorunsuz calisir.
- Hata durumlarinda tutarli donus mesaji uretilir.
- RAG akisi bozulmadan MCP ile birlikte calisir.
- Tum arac cagrilari loglardan izlenebilir durumdadir.

## 7. Faz Sonu Beklenen Cikti

Bu faz sonunda sistem:
- Araclarini MCP standardina tasimis olur
- Agent katmaninda entegrasyon karmasasini azaltir
- Guvenlik ve loglama disiplinini guclendirir

Bir sonraki adim: Degerlendirme seti ve production olgunlastirma.
